```python
##Do smoking, alcohol use, or obesity levels significantly influence cancer stage at diagnosis?
```

```python
df.head()
```

```python
unique_stages = df['Cancer_Stage'].unique()
```

```python
stage_order = {'Stage 0': 0, 'Stage I': 1, 'Stage II': 2, 'Stage III': 3, 'Stage IV': 4}
df['Cancer_Stage_Encoded'] = df['Cancer_Stage'].map(stage_order)
```

```python
correlations = df[['Cancer_Stage_Encoded', 'Smoking', 'Alcohol_Use', 'Obesity_Level']].corr()
correlations_with_stage = correlations['Cancer_Stage_Encoded'].drop('Cancer_Stage_Encoded')
```

```python
smoking_corr, smoking_p = stats.spearmanr(df['Cancer_Stage_Encoded'], df['Smoking'])
alcohol_corr, alcohol_p = stats.spearmanr(df['Cancer_Stage_Encoded'], df['Alcohol_Use'])
obesity_corr, obesity_p = stats.spearmanr(df['Cancer_Stage_Encoded'], df['Obesity_Level'])
```

```python
{"Correlations with Cancer Stage": correlations_with_stage.to_dict(),
    "Spearman P-Values": {
        "Smoking": smoking_p,
        "Alcohol_Use": alcohol_p,
        "Obesity_Level": obesity_p
    }
}
```

```python
##Do lifestyle factors (smoking, alcohol, obesity) vary significantly by region?
```

```python
df.head()
```

```python
grouped_smoking = [group['Smoking'].values for _, group in df.groupby('Country_Region')]
grouped_alcohol = [group['Alcohol_Use'].values for _, group in df.groupby('Country_Region')]
grouped_obesity = [group['Obesity_Level'].values for _, group in df.groupby('Country_Region')]

f_smoking, p_smoking = stats.f_oneway(*grouped_smoking)
f_alcohol, p_alcohol = stats.f_oneway(*grouped_alcohol)
f_obesity, p_obesity = stats.f_oneway(*grouped_obesity)

print("ANOVA Results:")
print(f"Smoking - F-stat: {f_smoking:.4f}, p-value: {p_smoking:.4f}")
print(f"Alcohol Use - F-stat: {f_alcohol:.4f}, p-value: {p_alcohol:.4f}")
print(f"Obesity Level - F-stat: {f_obesity:.4f}, p-value: {p_obesity:.4f}")
```
